# Tests de plusieurs scénarios d'interaction — chaîne RAG complète

Teste `scripts/rag_chain.py` (retrieval + génération, pas juste la recherche comme à l'Étape 3) sur plusieurs questions représentatives, pour vérifier la demande explicite de l'Étape 4 ("Tester plusieurs scénarios d'interaction pour vérifier la robustesse du chatbot").

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path("..") / "scripts"))

from rag_chain import build_chain

chain = build_chain()
print("Chaîne RAG construite.")
```

```python
def ask(question: str) -> None:
    print(f"### Question : {question}\n")
    print(chain.invoke(question))
    print()
```

## Scénario 1 — recherche thématique + lieu

```python
ask("Quels concerts de musique classique y a-t-il à Marseille ?")
```

## Scénario 2 — contrainte de prix

```python
ask("Je cherche un événement gratuit pour sortir avec mes enfants.")
```

## Scénario 3 — question multi-contraintes (thème + budget + public)

```python
ask("Un événement romantique pour un couple, pas trop cher, plutôt en soirée ?")
```

## Scénario 4 — vérifier qu'un événement annulé est signalé comme tel dans la réponse

Pas juste retrouvé (déjà vérifié à l'Étape 3) — ici on vérifie que le LLM **mentionne explicitement** l'annulation dans sa réponse rédigée, plutôt que de le passer sous silence.

```python
ask("L'atelier d'écriture des bibliothèques de Marseille est-il maintenu ?")
```

## Scénario 5 — cas limite : question hors-sujet

Vérifie que la consigne système ("si aucun événement ne correspond, dis-le clairement") fonctionne réellement, plutôt que de forcer une réponse à partir de résultats non pertinents.

```python
ask("Quelle est la recette d'un bon gâteau au chocolat ?")
```

## Scénario 6 — question ambiguë/vague

```python
ask("Qu'est-ce qu'il y a à faire ce week-end ?")
```

## Conclusion

**Scénario 1 (thème + lieu)** — Pertinent et honnête : l'événement retrouvé (concert à la bougie) est bien à Marseille, correspond au thème demandé, et la date affichée est correcte et vérifiée (mardi 1er septembre 2026, aujourd'hui).

**Scénario 2 (contrainte de prix)** — Deux événements gratuits, cohérents avec la demande (sortie en famille), dates à venir vérifiées correctes (mercredi 9 et samedi 19 septembre).

**Scénario 3 (multi-contraintes)** — Réponse pertinente et bien justifiée (choix expliqué), date et heure vérifiées correctes après le fix des événements récurrents (19h00, heure locale — voir points techniques ci-dessous).

**Scénario 4 (honnêteté sans résultat)** — Comportement volontairement modifié en cours de projet : plutôt que de signaler un atelier annulé (comportement initial), le filtre de date par défaut exclut désormais tout événement déjà terminé, y compris pour les questions sans période explicite. Le chatbot répond donc honnêtement qu'aucun atelier d'écriture n'est disponible, sans forcer de rapprochement avec un événement non pertinent.

**Scénario 5 (hors-sujet)** — Refus correct et honnête (pas de recette de cuisine), tout en restant utile : propose des ateliers culinaires réels et à venir, sans sortir du périmètre de Puls-Events.

**Scénario 6 (question temporelle vague)** — "Ce week-end" est correctement borné aux bonnes dates (5-6 septembre 2026), cohérent avec le Scénario 3 sur l'événement commun.

### Points techniques validés au fil des tests

- **Recherche hybride** (filtre structuré + sémantique) : corrige le bug initial où une recherche purement sémantique ne pouvait pas exclure les événements passés ni borner une période demandée.
- **Chevauchement de dates** (`date_start`/`date_end`) : un événement déjà commencé mais encore en cours reste inclus, au lieu d'être exclu à tort sur la seule base de sa date de début.
- **Filtre de date par défaut** (`date_end >= aujourd'hui`) : appliqué même sans période explicite dans la question — un événement terminé n'est jamais une réponse valable, et ce tri ne peut pas être laissé au jugement du LLM (constaté peu fiable empiriquement, y compris avec la date du jour fournie dans le prompt).
- **Événements récurrents** (`occurrences`) : la date affichée est désormais la prochaine occurrence réelle, pas systématiquement la première (potentiellement déjà passée) ; un second contrôle (`occurrence_in_period`) vérifie qu'une occurrence tombe bien dans la période demandée, au-delà du simple chevauchement `date_start`/`date_end`.
- **Effet de bord positif** : ce même champ `occurrences` (stocké en heure locale) a aussi corrigé un décalage silencieux de 2h sur l'affichage des heures, présent depuis le début de l'Étape 4 (`date_start` était stocké en UTC, jamais reconverti à l'affichage).
- **Jour de la semaine** : calculé en Python (`format_date_fr`) plutôt que déduit par le LLM à partir d'une date ISO brute, après une incohérence observée entre deux réponses sur le même événement.

### Limite connue, non traitée à ce stade

La chaîne est **stateless** : chaque question est traitée indépendamment, sans mémoire des échanges précédents. Une question de suivi sans contexte explicite (ex. "je veux l'url de l'événement" après une première question) n'a aujourd'hui aucun moyen d'être résolue correctement — nécessiterait une gestion d'historique de conversation.
